The functions below will download the dataset into your browser:



```python
# Define the URL of the dataset
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### 2. Explore the Dataset
##### Task 1: Display basic information and summary statistics of the dataset.


```python
## Write your code here
df.info()
```

```python
df.describe(include='all')
```

### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
## Write your code here
df.isnull().sum()
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
## Write your code here

plt.figure(figsize=(12, 8))
sns.heatmap(df.isnull(), yticklabels=False, cbar=False, cmap='viridis')

plt.title('Missing Values Heatmap')
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
## Write your code here
missing_employment = df['Employment'].isnull().sum()

print(f"Number of missing rows in Employment: {missing_employment}")
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
## Write your code here
majority_employment = df['Employment'].mode()[0]

print(f"The most frequent value in Employment is: {majority_employment}")
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
## Write your code here
df['Employment'] = df['Employment'].fillna(majority_employment)

print(f"Missing values remaining in Employment: {df['Employment'].isnull().sum()}")
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
## Write your code here

plt.figure(figsize=(12, 12))
sns.countplot(
    data=df, 
    y='Employment', 
    hue='Employment', 
    palette='viridis', 
    legend=False,
    order=df['Employment'].value_counts().index
)

plt.title('Distribution of Employment After Imputation', fontsize=15)
# plt.xticks(rotation=90) # Rotate labels if they overlap
plt.show()
```
