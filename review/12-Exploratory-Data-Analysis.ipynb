#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
## Write your code here
critical_cols = ['Employment', 'JobSat', 'RemoteWork']

print(df[critical_cols].isnull().sum())
print(df[critical_cols].isnull().mean() * 100) # percentage of missing 
```

```python
# fill with a new category
df['JobSat'] = df['JobSat'].fillna('Missing')
df['RemoteWork'] = df['RemoteWork'].fillna('Unknown')
```

```python
# option 2: impute with the most frequent value
for col in ['Employment', 'RemoteWork']:
    df[col] = df[col].fillna(df[col].mode()[0])

# option 3: drop missing rows
df.dropna(subset=['Employment', 'JobSat', 'RemoteWork'], inplace=True)
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
df['YearsCodePro'].info
#df['JobSat'].info
```

```python
## Write your code here
# Convert to numeric, forcing non-numbers to NaN 
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df['YearsCodePro'].info
```

```python
df = df.dropna(subset=['YearsCodePro'])
df['YearsCodePro'].info
```

```python
bins = [0, 5, 10, 20, 100]
labels = ['0-5','5-10','10-20','20+']
df['ExpRange'] = pd.cut(df['YearsCodePro'], bins=bins, labels=labels)
```

```python
print(df['JobSat'].unique())
```

```python
df['JobSatNumeric'] = pd.to_numeric(df['JobSat'], errors='coerce')
df_clean = df.dropna(subset=['JobSatNumeric', 'ExpRange'])
```

```python
median_sat = df.groupby('ExpRange')['JobSatNumeric'].median() #.reset_index()

plt.figure(figsize=(10, 6))
#sns.barplot(data=df_clean, x='ExpRange', y='JobSatNumeric', estimator='median', errorbar=None)
sns.barplot(
    x=median_sat.index,
    y=median_sat.values,
    hue=median_sat.values,
    palette='Blues_d'
)
plt.title('Median Job Satisfaction by Professional Experience')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Median Satisfaction (0-10 Scale)')
plt.ylim(0, 10)
plt.show()
```

```python
# Calculate the mean instead of the median
plt.figure(figsize=(10, 6))
mean_stats = df_clean.groupby('ExpRange', observed=True)['JobSatNumeric'].mean().round(2)

sns.barplot(
    x=mean_stats.index,
    y=mean_stats.values,
    hue=mean_stats.values,
    palette='Blues_d'
)

plt.title('Mean Job Satisfaction by Professional Experience')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Mean Satisfaction Score (0-10 Scale)')
plt.ylim(0, 10) 
plt.show()

print(df.groupby('ExpRange', observed=True)['JobSatNumeric'].mean())
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
## Write your code here
df['LanguageHaveWorkedWith'].info
```

```python
df['Country'].info
```

```python
target_regions = ['United States of America', 'India', 'Germany', 'United Kingdom of Great Britain and Northern Ireland']
df_filtered = df[df['Country'].isin(target_regions)].copy()
```

```python
df_filtered['LanguageHaveWorkedWith'] = df_filtered['LanguageHaveWorkedWith'].str.split(';')
df_lang = df_filtered.explode('LanguageHaveWorkedWith')
```

```python
top_langs = df_lang.groupby(['Country', 'LanguageHaveWorkedWith']).size().reset_index(name='Count')
top_langs = top_langs.sort_values(['Country', 'Count'], ascending=[True, False]).groupby('Country').head(10)
```

```python
plt.figure(figsize=(14, 8))
sns.barplot(data=top_langs, x='Count', y='LanguageHaveWorkedWith', hue='Country')
plt.title('Top 10 Programming Languages by Region')
plt.show()
```

```python
df_lang.reset_index(drop=True, inplace=True)

top_20_global = df_lang['LanguageHaveWorkedWith'].value_counts().nlargest(20).index

df_top_20 = df_lang[df_lang['LanguageHaveWorkedWith'].isin(top_20_global)]

pivot_df = pd.crosstab(df_top_20['Country'], df_top_20['LanguageHaveWorkedWith'], normalize='index')

# 5. Plot the heatmap
plt.figure(figsize=(16, 10))
sns.heatmap(pivot_df.T, annot=True, cmap='YlGnBu', fmt='.1%')

plt.title('Usage Intensity of Top 20 Global Programming Languages by Region')
#plt.xlabel('Country')
#plt.ylabel('Programming Language')
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
## Write your code here
df['JobSatPoints_1'].unique()
```

```python
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df['JobSatPoints_1'] = pd.to_numeric(df['JobSatPoints_1'], errors='coerce')
```

```python
df_plot = df.dropna(subset=['YearsCodePro', 'JobSatPoints_1'])
```

```python
plt.figure(figsize=(10, 6))
# Using alpha=0.5 helps manage overlapping points if the dataset is large
sns.scatterplot(x='YearsCodePro', y='JobSatPoints_1', data=df_plot, alpha=0.5)

plt.title('Relationship Between Years of Experience and Job Satisfaction')
plt.xlabel('Professional Coding Experience (Years)')
plt.ylabel('Job Satisfaction Score')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.regplot(x='YearsCodePro', y='JobSatPoints_1', data=df_plot, 
            scatter_kws={'alpha':0.3}, line_kws={'color':'red'})

correlation = df_plot['YearsCodePro'].corr(df_plot['JobSatPoints_1'])

plt.title(f'Experience vs. Satisfaction (Correlation: {correlation:.2f})')
plt.xlabel('Professional Coding Experience (Years)')
plt.ylabel('Job Satisfaction Score')
plt.show()

print(f"The Pearson correlation coefficient is: {correlation:.4f}")
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
## Write your code here
df_clean = df.dropna(subset=['EdLevel', 'Employment']).copy()
df_clean['PrimaryEmployment'] = df_clean['Employment'].str.split(';').str[0]
```
