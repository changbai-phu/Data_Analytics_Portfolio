### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
## Write your code here
job_sat_counts = df['JobSat'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(job_sat_counts, labels=job_sat_counts.index, autopct='%1.1f%%', 
        startangle=140, colors=sns.color_palette('viridis'))

plt.title('Distribution of Job Satisfaction Levels')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(df['JobSatPoints_1'], fill=True, color="dodgerblue", bw_adjust=0.5)

plt.title('KDE Plot of Job Satisfaction Scores')
plt.xlabel('Satisfaction Score (0-100)')
plt.ylabel('Density')
plt.show()
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
## Write your code here
have_lang = df['LanguageHaveWorkedWith'].str.split(';').explode()
have_counts = have_lang.value_counts().reset_index()
have_counts.columns = ['Language', 'Count']
have_counts['Type'] = 'Have Worked With'

want_lang = df['LanguageWantToWorkWith'].str.split(';').explode()
want_counts = want_lang.value_counts().reset_index()
want_counts.columns = ['Language', 'Count']
want_counts['Type'] = 'Want to Work With'

combined = pd.concat([have_counts, want_counts])
top_10 = have_counts.nlargest(10, 'Count')['Language']
combined_filtered = combined[combined['Language'].isin(top_10)]
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=combined_filtered, x='Language', y='Count', hue='Type')
plt.title('Usage vs. Desire: Top 10 Programming Languages')
plt.xticks(rotation=45)
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
## Write your code here
top_countries = df['Country'].value_counts().nlargest(10).index
df_filtered = df[df['Country'].isin(top_countries)]

plt.figure(figsize=(14, 7))
sns.countplot(data=df_filtered, y='Country', hue='RemoteWork')

plt.title('Remote Work Distribution by Top Countries')
#plt.xticks(rotation=45)
#plt.ylabel('Number of Respondents')
plt.legend(title='Work Status', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
## Write your code here
df['JobSat_Numeric'] = pd.to_numeric(df['JobSat'], errors='coerce')
df['YearsCodePro_Numeric'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df_corr = df.dropna(subset=['JobSat_Numeric', 'YearsCodePro_Numeric'])
```

```python
!pip install scipy
```

```python
pearson_corr = df_corr['YearsCodePro_Numeric'].corr(df_corr['JobSat_Numeric'], method='pearson')
spearman_corr = df_corr['YearsCodePro_Numeric'].corr(df_corr['JobSat_Numeric'], method='spearman')

print(f"Pearson Correlation: {pearson_corr:.4f}")
print(f"Spearman Correlation: {spearman_corr:.4f}")
```

```python
Weak Correlation: In developer datasets, these values typically fall between -0.1 and 0.1, indicating that years of experience have almost no direct linear impact on how satisfied a person is with their job.
Plateau Effect: Satisfaction is often more influenced by salary, management quality, and work-life balance rather than simply the length of one's career.
Non-Linearity: If the Spearman coefficient is higher than the Pearson, it suggests a trend exists that isn't a straight line (e.g., satisfaction might rise in the first 5 years and then level off).
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
## Write your code here
df['PrimaryEmployment'] = df['Employment'].str.split(';').str[0]
df_clean = df.dropna(subset=['EdLevel', 'PrimaryEmployment'])

ct = pd.crosstab(df_clean['EdLevel'], df_clean['PrimaryEmployment'], normalize='index')
```

```python
ax = ct.plot(kind='bar', stacked=True, figsize=(12, 7))
plt.title('Employment Status by Education Level')
plt.xlabel('Education Level (EdLevel)')
plt.ylabel('Proportion of Respondents')
plt.legend(title='Employment Status', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
